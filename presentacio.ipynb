# Presentació de les pràctiques amb ordinador

Les pràctiques amb ordinador corresponen als continguts del tema 14, sobre la humanitat actual. Són 5 sessions en aula d'informàtica, amb els continguts següents:

- Pràctica 1: La teoria de la coalescència.
- Pràctica 2: Selecció i deriva.
- Pràctica 3: Filogènia dels hominins.
- Pràctica 4: El rellotge molecular.
- Pràctica 5: Mestissatge en poblacions actuals.

L'assistència a les sessions de pràctiques amb ordinador és obligatòria. Es passarà llista. A més, a l'examen apareixeran un parell de preguntes sobre les pràctiques. No cal lliurar cap quadern de pràctiques ni cap treball sobre les mateixes.

Totes les pràctiques es realitzaran mitjançant l'execució de comandaments del llenguatge de programació R, en l'entorn de quaderns [Jupyter](https://jupyter.org/). Podem editar i executar els quaderns a través del navegador d'internet des dels ordinadors virtuals de la Universitat de València amb el sistema operatiu Ubuntu, on Jupyter i la resta de paquets i programes ja estan instal·lats. Podeu iniciar un ordinador virtual amb Ubuntu a través de l'enllaç següent (utilitzeu VPN des de fora de la UV):

[escritorio.uv.es](escritorio.uv.es)

Si voleu instal·lar Jupyter i executar les pràctiques als vostres ordinadors, podeu seguir les instruccions [en aquest enllaç](https://jupyter.org/install). Tingueu en compte que necessitareu tenir Python instal·lat. A més, a les pràctiques utilitzarem R i els paquets d'R següents, que també cal tenir instal·lats si voleu utilitzar els vostres propis ordinadors: `learnPopGen`, `phangorn`, `DECIPHER` i `pegas`. Per últim, a la pràctica 5 utilitzarem els programes `admixture`, `plink` (1.9) i `evalAdmix` (trobareu referències i instruccions d'instal·lació al guió de la pràctica 5). Tots aquests programes i paquets són programari lliure. Els guions de pràctiques i els materials adicionals també estan disponibles amb la llicència *Attribution-ShareAlike 4.0 International* de *Creative Commons* en l'adreça següent: [github.com/IgnasiLucas/PTE_33056](github.com/IgnasiLucas/PTE_33056).
