## El model
Les figures com l'anterior assumeixen un model demogràfic concret, amb les assumpcions següents:
1. Les poblacions ancestrals van viure separades, van divergir per deriva genètica i poden haver-se mesclat recentment.
2. Les poblacions mestisses no han patit deriva genètica des del mestissatge: la seua constitució genètica s'explica adequadament per les proporcions de mestissatge i les freqüències al·lèliques inicials.
3. Totes les poblacions ancestrals estan ben representades en la mostra.

## Les dades
Les dades de partida són:

1. Una gran matriu **G**, dels genotips en *M* loci (normalment, milers d'SNPs) d'*N* individus;
2. i un número de poblacions ancestrals que suposem que van existir, *K*.

Friedlaender et al. (2008) van utilitzar dades genètiques de 890 loci en 1893 individus de 91 poblacions diferents. Nosaltres utilitzarem els genotips de 180 persones en més de 54000 loci (amb només 0.8% de dades absents). Els 180 individus són 30 de la península Ibèrica (IBS), 30 italians de la Toscana (TSI), 30 colombians de Medellín (CLM), 30 peruans de Lima (PEL), 30 mexicans de Los Angeles (MXL) i 30 japonesos de Tòkio (JPT). Aquestes dades provenen de la re-seqüenciació a alta cobertura de més de 3000 individus participants del projecte dels 1000 genomes [(Byrska-Bishop et al. 2022)](https://doi.org/10.1016/j.cell.2022.08.004).

Les dades estan a la carpeta `data`, en un format binari propi del programa `plink` [(Chang et al. 2015)](https://doi.org/10.1186/s13742-015-0047-8). Per poder transformar-les a un format de text pla, cal tenir instal·lat el programa `plink`.

Des d'aquest quadern Jupyter només podem executar ordres del llenguatge R. Per executar comandaments del sistema, els hem d'embolicar amb la funció `system2()` d'R.

El bloc següent generarà una versió llegible de les dades genètiques (la matriu **G**), en l'arxiu `data/humans.ped`:

```r
if (! file.exists('data/humans.ped')) 
   system2('plink', args = c('--bfile', 'data/humans',
                             '--out', 'data/humans',
                             '--recode'),
          stdout = FALSE, stderr = FALSE)
```

## Mètodes
El programa `admixture` utilitza la matriu **G** i el número de poblacions ancestrals *K* per *estimar* dues matrius: la matriu **P** de les freqüències al·lèliques dels *M* loci en cadascuna de les *K* poblacions ancestrals; i la matriu **Q** ($N\times K$), de les proporcions de loci que cada individu ha heretat de cada una de les *K* poblacions. El bloc de codi següent realitza l'estimació per a $K=2$. L'opció `--cv` (*Cross Validation*) instrueix el programa per estimar també una taxa d'error en l'assignació d'ancestralitats.

```r
if (! file.exists('humans.2.Q'))   # Esta línia no cal copiar-la en l'exercici 4.
    system2('admixture', args = c('--cv', 'data/humans.bed', '2'),
           stdout = 'humans.2.log')
```

## Resultats
És hora de visualitzar els resultats. Explora alguna de les matrius de proporcions de mestissatge, **Q**. Voràs que cada matriu té tantes files com individus i tantes columnes com poblacions ancestrals hem assumit. Les files sumen 1, perquè cada columna representa la proporció estimada del genoma que cada individu ha heretat de cada població ancestral. El codi següent representa cada fila de **Q** com una barra vertical amb fragments de colors diferents proporcionals a la fracció del genoma heretada de cada suposada població ancestral.

```r
Q2 <- matrix(scan('humans.2.Q'), ncol = 2, byrow = TRUE)
options(repr.plot.width = 21, repr.plot.height = 7, repr.plot.res = 100)
barplot(t(Q2), col = c(2,3))
```

### Exercici 5. En blocs de codi addicionals, genera gràfics com l'anterior per a cada valor de *K*. Fixa't en tot el que has de canviar!

Un aspecte important és que el mètode utilitzat per `admixture` ignora per complet les *etiquetes* que podem haver assignat als individus. El coneixement previ sobre l'origen de les mostres pot utilitzar-se per interpretar els resultats. Els individus (és a dir, les files en la matriu **Q**) estan disposats en el mateix ordre en què apareixen els individus en la matriu **G** de genotips. La identitat dels individus podem trobar-la en les primeres columnes d'aquella matriu **G** (és a dir, a l'arxiu `data/humans.ped`) o, per més comoditat, en l'arxiu `data/humans.fam`. Obri aquest últim i observa les dues primeres columnes, que corresponen a una població assignada i un identificador únic d'individu. Podem afegir esta informació als gràfics anteriors amb les instruccions següents:

```r
metadata <- read.table('data/humans.fam', col.names = c('pop', 'ind', 'dad', 'mom', 'sex', 'phe'),
                       colClasses = c('factor', 'character', 'NULL', 'NULL', 'NULL', 'NULL'))
row.names(Q2) <- metadata$ind
# "space=0" anula l'espai entre barres, fent que la posició de cada individu siga igual al seu número d'ordre.
# "las=2" fa que els noms dels individus s'escriguen perpendiculars a l'eix.
barplot(t(Q2), col = c(2, 3), space = 0, las = 2)
popPosition <- sapply(split(1:dim(metadata)[1], metadata$pop), mean)
popMaximum  <- sapply(split(1:dim(metadata)[1], metadata$pop), max)
axis(3, at = popPosition, labels = names(popPosition), tick = 0)
abline(v = popMaximum, lwd = 2, col = 'white')
```

## Avaluació dels resultats
El programa `admixture` ofereix un mètode per avaluar quin valor de *K* és més adequat per a les dades. Consisteix en identificar el valor de *K* amb un *error de validació creauada* menor. Aquests errors es troben al final dels arxius `.log`.

### Exercici 7. Ompli el vector `CVerrors` del bloc següent manualment amb els valors que trobaràs als arxius `.log`, i després executa el bloc per obtenir-ne una representació gràfica.

```r
CVerrors <- c(0, 0, 0, 0, 0)
K <- c(2, 3, 4, 5, 6)
options(repr.plot.width = 7, repr.plot.height = 7, repr.plot.res = 100)
plot(K, CVerrors, type = 'b')
```

Els errors de validació creuada ens indiquen quin valor de *K* és més adequat. Però aquest mètode no qüestiona el model demogràfic assumit pel programa, a saber, que les poblacions provenen de *K* poblacions ancestrals, que havien divergit en aïllament i que poden haver-se mesclat només recentment. El programa `evalAdmix` (Garcia-Erill i Albrechtsen, 2020) comprova si les dades s'ajusten a aquest model. La lògica d'aquest programa és que si els genotips de tots els individus poden ser adequadament explicats pel model, les diferències entre els genotips observats i els que prediu el model per a un individu (els *residuals*) no tenen per què estar correlacionades amb els residuals de cap altre individu. Els blocs de codi següents calculen i representen la matriu de correlacions entre residuals de parelles d'individus amb $K=2$.

```r
source('visFuns.R')
pop <- read.table('data/humans.fam')
palette(c("#A6CEE3", "#1F78B4", "#B2DF8A", "#33A02C", "#FB9A99", "#E31A1C", "#FDBF6F",
          "#FF7F00", "#CAB2D6", "#6A3D9A", "#FFFF99", "#B15928", "#1B9E77", "#999999"))
```

```r
K <- 2
EvalFile <- paste0('Eval.', K, '.txt')
if (! file.exists(EvalFile)) {
   system2('evalAdmix', 
           args = c('-plink', 'data/humans', 
                    '-fname', paste0('humans.', K, '.P'),
                    '-qname', paste0('humans.', K, '.Q'),
                    '-o', EvalFile, '-P', '1'),
           stdout = paste0('Eval.', K, '.log'))
}
q <- read.table(paste0('humans.', K, '.Q'), stringsAsFactors = TRUE)
ord <- orderInds(pop = as.vector(pop[,1]), q = q)
r <- as.matrix(read.table(EvalFile))
plotCorRes(cor_mat = r, pop = as.vector(pop[, 1]), ord = ord,
           title = paste0('Evaluation of admixture proportions with K =', K),
           max_z = 0.5, min_z = -0.5)
```
